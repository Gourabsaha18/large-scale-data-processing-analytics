# 02 - PySpark ETL\nRun the scalable cleaning and transformation workflow.

```python
from pyspark.sql import SparkSession\nfrom pyspark.sql.functions import col, year, month\n\nspark = SparkSession.builder.appName('SalesETL').getOrCreate()
```

```python
df = spark.read.option('header', True).option('inferSchema', True).csv('../data/raw/sales.csv')\ndf.printSchema()\nprint(df.count())
```

## Cleaning and transformations\nAdapt column names to the selected dataset.

```python
df = df.dropDuplicates().dropna(how='all')\nif 'Quantity' in df.columns and 'Unit_Price' in df.columns:\n    df = df.withColumn('Revenue', col('Quantity') * col('Unit_Price'))\n\ndf.show(5)
```

```python
df.write.mode('overwrite').parquet('../data/processed/sales_parquet')
```
