# 03 - Spark SQL Analysis

```python
from pyspark.sql import SparkSession\nspark = SparkSession.builder.appName('SalesSQL').getOrCreate()\ndf = spark.read.parquet('../data/processed/sales_parquet')\ndf.createOrReplaceTempView('sales')
```

```python
spark.sql('''SELECT Category, ROUND(SUM(Revenue),2) AS revenue FROM sales GROUP BY Category ORDER BY revenue DESC''').show()
```

```python
spark.sql('''SELECT Region, ROUND(SUM(Revenue),2) AS revenue FROM sales GROUP BY Region ORDER BY revenue DESC''').show()
```

```python
spark.sql('''SELECT Year, Month, ROUND(SUM(Revenue),2) AS revenue FROM sales GROUP BY Year, Month ORDER BY Year, Month''').show()
```
